# Step 0 · Hello, Ollama
The simplest thing possible: send one question to a model running **on this machine**, and read the answer.

No gateway. No tracing. No app code. Only Ollama, which must be running (`ollama serve`, or open the Ollama app).

## 1. Is Ollama running?
Ollama is a small web server on your own machine, at port 11434. This asks it which models it has.

```python
import requests

OLLAMA_URL = "http://localhost:11434"
MODEL = "qwen2.5:7b-instruct"

models = requests.get(f"{OLLAMA_URL}/api/tags", timeout=10).json()["models"]
for model in models:
    print(f"{model['name']:<25} {model['size'] / 1e9:.1f} GB")
```

## 2. Ask it one question

```python
answer = requests.post(f"{OLLAMA_URL}/api/chat", timeout=600, json={
    "model": MODEL,
    "messages": [{"role": "user", "content": "In one sentence: what does a payroll provider do for its clients?"}],
    "stream": False,
}).json()

print(answer["message"]["content"])
```

## 3. How fast was that?
Ollama reports its own timings in the same reply, in nanoseconds.

```python
tokens = answer["eval_count"]
seconds = answer["eval_duration"] / 1e9
print(f"{tokens} tokens in {seconds:.1f} seconds = {tokens / seconds:.1f} tokens per second")
print(f"Total time, including loading the model: {answer['total_duration'] / 1e9:.1f} seconds")
```

**Takeaway:** a local model is one web request away, and nothing left this machine.

**What this does not give us:** anything stops the code calling any model it likes, and no record exists of what was asked.
